### Task 1: Calculate Tuition at UIC

Write a program that determine's the student's **yearly tuition bill** at UIC. One year of tuition includes two semesters. Table 1: https://uofi.app.box.com/s/3ub3lw8qeda73k7u9bl2927o66vhyvzq 

This program should ask tbe user to input: 
1. the student's `residential_status`: in-state or out-of-state.
2. The amount of `credits` the student is registered for.
3. Whether the student has a `late` payment notice on their account, and if so the `unpaid_fees` (float), and for how many `months` this notice has been active

Late payments are subject to an additional charge of $1.00 or 1.5% of their semester bill (whichever is greater). AND a penalty of $2.00 per month.

Below are the rules this program should follow:
- If the number of `credits` is 0, then the bill is equal to $971 plus any `unpaid_fees`.

For in-state students
- If the number of `credits` is between 1 and 5, then each semester's fee is $1942 plus any `unpaid_fees`. 
- If the number of `credits` is between 6 and 11, then each semester's fee is $3884 plus any `unpaid_fees`. 
- If the number of `credits` is 12 or more, then each semester's fee is $5826 plus any `unpaid_fees`. 

For out-of-state students
- If the number of `credits` is between 1 and 5, then each semester's fee is $4725 plus any `unpaid_fees`. 
- If the number of `credits` is between 6 and 11, then each semester's fee is $9451 plus any `unpaid_fees`. 
- If the number of `credits` is 12 or more, then each semester's fee is $14176 plus any `unpaid_fees`. 

Lastly, this program should print the final amount of money the student owes..

## Important Considerations
1. The number of `credits` should be greater than 0.
2. If the `unpaid_fees` are less than 0, that amount counts as a credit to the students bill and should be subtracted.
3. The bill should be displayed.
4. The code should avoid redundancies within the statements.

```python
#Task 1 Program


residential_status = input("Enter residential status: ")
credits = int(input("Enter credits: "))
late = input("Late payment notice? yes or no: ")

unpaid_fees = 0
months = 0

if late == "yes":
    unpaid_fees = float(input("Enter unpaid fees: "))
    months = int(input("Enter number of months: "))

# Find semester tuition
if credits == 0:
    tuition = 971

elif residential_status == "in-state":
    if credits <= 5:
        tuition = 1942
    elif credits <= 11:
        tuition = 3884
    else:
        tuition = 5826

else:
    if credits <= 5:
        tuition = 4725
    elif credits <= 11:
        tuition = 9451
    else:
        tuition = 14176

# Add unpaid fees
tuition += unpaid_fees

# Add late fees
if late == "yes":
    late_fee = tuition * 0.015

    if late_fee < 1:
        late_fee = 1

    tuition = tuition + late_fee + (months * 2)

# Two semesters in one year
yearly_bill = tuition * 2

print("Yearly tuition bill: $", yearly_bill)

```
